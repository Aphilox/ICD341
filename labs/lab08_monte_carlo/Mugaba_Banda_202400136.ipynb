### Questions 6
Use importance sampling to estimate 𝐸(𝑋2) where 𝑋~𝑈(0,1). 

```python
np.random.seed(4)
n = 100_000

# True value: E[X^2] for X ~ U(0,1) is 1/3
# Simple MC
x = np.random.uniform(0, 1, n)
mc_estimate = np.mean(x**2)
print("Simple MC estimate:", mc_estimate)

# Importance sampling with a beta(2,1) proposal (density f(x) = 2x on [0,1])
x_is = np.random.beta(2, 1, n)
# weights: target density / proposal density = 1 / (2x)
weights = 1 / (2 * x_is)
is_estimate = np.mean((x_is ** 2) * weights)
print("Importance sampling estimate:", is_estimate)
print("True value: 1/3 =", 1/3)
```

### Questions 7
 Explain the difference between importance sampling and control variates.

**Difference between importance sampling and control variates**

- **Importance sampling:** We sample from a different distribution (the "proposal" distribution) that puts more weight on important regions of the target. We then correct for the bias using a likelihood ratio (weight).

- **Control variates:** We keep the original sampling distribution but subtract a correlated random variable with a known mean, reducing variance. It doesn't change the sampling distribution—only adjusts the estimator.

In short: importance sampling changes *where* we sample; control variates change *what we add/subtract* from the estimator.
