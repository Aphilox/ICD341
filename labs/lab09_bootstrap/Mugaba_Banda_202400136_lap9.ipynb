```python
import numpy as np

data = np.array([3, 4, 5, 7, 8, 9])
n_iterations = 10000
boot_means = []

for _ in range(n_iterations):
    # Sample with replacement
    sample = np.random.choice(data, size=len(data), replace=True)
    boot_means.append(np.mean(sample))

# 95% percentile interval
lower = np.percentile(boot_means, 2.5)
upper = np.percentile(boot_means, 97.5)

print(f"95% CI for the mean: [{lower:.2f}, {upper:.2f}]")
```

```python
boot_medians = []

for _ in range(n_iterations):
    sample = np.random.choice(data, size=len(data), replace=True)
    boot_medians.append(np.median(sample))

lower_med = np.percentile(boot_medians, 2.5)
upper_med = np.percentile(boot_medians, 97.5)

print(f"95% CI for the median: [{lower_med:.2f}, {upper_med:.2f}]")
```

```python
# Fit a normal model to the data
mu = np.mean(data)
sigma = np.std(data, ddof=1)

param_boot_means = []
for _ in range(n_iterations):
    # Generate data from the fitted normal distribution
    sample = np.random.normal(mu, sigma, size=len(data))
    param_boot_means.append(np.mean(sample))

param_lower = np.percentile(param_boot_means, 2.5)
param_upper = np.percentile(param_boot_means, 97.5)

print(f"Parametric 95% CI for the mean: [{param_lower:.2f}, {param_upper:.2f}]")
```

    **Q4: Data leakage during model validation**
    Data leakage occurs when information from the test set is accidentally used during the training phase. One way this happens is **preprocessing before splitting**: if you scale (normalize) the entire dataset before splitting it into train and test sets, the mean and standard deviation used for scaling will include information from the test set. The model will look artificially good on the test set because it "peeked" at the answers during training.
