### Questions 1
Create a covariance matrix and perform Cholesky decomposition. 

```python
import numpy as np

# Create some sample data
np.random.seed(42)
data = np.random.randn(100, 4)

# Covariance matrix (must be symmetric positive-definite for Cholesky)
cov_matrix = np.cov(data, rowvar=False)
print("Covariance matrix:\n", cov_matrix)

# Cholesky decomposition: cov_matrix = L @ L.T
L = np.linalg.cholesky(cov_matrix)
print("\nCholesky factor L:\n", L)

# Verify
print("\nVerify L @ L.T equals cov_matrix:")
print(np.allclose(L @ L.T, cov_matrix))
```

### Questions 2
 Perform QR decomposition on a matrix. 

```python
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

Q, R = np.linalg.qr(A)
print("Q:\n", Q)
print("\nR:\n", R)

# Verify
print("\nVerify Q @ R equals A:")
print(np.allclose(Q @ R, A))
```

### Questions 3
 Reconstruct a matrix using SVD. 

```python
A = np.array([[1, 2, 3],
              [4, 5, 6],
              [7, 8, 10]])

U, S, Vt = np.linalg.svd(A)

# Reconstruct using the formula: A = U @ diag(S) @ Vt
S_matrix = np.diag(S)
A_reconstructed = U @ S_matrix @ Vt

print("Original:\n", A)
print("\nReconstructed:\n", A_reconstructed)
print("\nVerify:", np.allclose(A, A_reconstructed))
```

### Questions 4
 Calculate the condition number of a matrix.

```python
A = np.array([[1, 2],
              [3, 4.0001]])  # Nearly singular

cond = np.linalg.cond(A)
print("Condition number:", cond)
```

**Condition number interpretation**

The condition number measures how sensitive the solution of a linear system is to small changes in the input. A very large condition number (e.g., \(10^4\) or more) means the matrix is **ill-conditioned** — small errors in the data can cause huge errors in the solution.

### Questions 5
 Create a sparse 1000 × 1000 matrix.

```python
from scipy.sparse import csr_matrix
import numpy as np

# Create a sparse 1000x1000 matrix with only a few non-zero entries
rows = np.random.randint(0, 1000, size=100)
cols = np.random.randint(0, 1000, size=100)
vals = np.random.rand(100)

sparse_mat = csr_matrix((vals, (rows, cols)), shape=(1000, 1000))

print("Shape:", sparse_mat.shape)
print("Non-zero entries:", sparse_mat.nnz)
print("Memory used by data (bytes):", sparse_mat.data.nbytes)
print("Memory used by indices (bytes):", sparse_mat.indices.nbytes)
```

### Questions 6
Explain why sparse matrix are useful for large datasets

**Sparse matrices are useful because:*

1. **Memory efficiency:** They store only the non-zero elements and their indices, rather than storing thousands or millions of zeros.
2. **Speed:** Operations skip over zeros entirely, so they run much faster.
3. **Scalability:** For real-world data (text documents, social networks, recommender systems), matrices are often 99%+ zeros. Dense storage would be impossible.

Examples: term-document matrices in NLP, adjacency matrices in graph problems, and user-item matrices in recommendation systems.
