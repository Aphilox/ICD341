```python
from numba import njit

@njit
def sum_squares_numba(x):
    total = 0.0
    for val in x:
        total += val**2
    return total

print("Sum of squares (Numba):", sum_squares_numba(x))
```

```python
from joblib import Parallel, delayed
import numpy as np

def bootstrap_mean(data):
    sample = np.random.choice(data, size=len(data), replace=True)
    return np.mean(sample)

data = np.random.randn(100)

# Run 1000 bootstraps in parallel across all CPU cores (-1)
results = Parallel(n_jobs=-1)(
    delayed(bootstrap_mean)(data) for _ in range(1000)
)

print(f"Mean of 1000 bootstrapped means: {np.mean(results):.4f}")
```

```python
import dask.array as da

# Create a massive random array (1 billion elements) chunked into 100M pieces
x_dask = da.random.random((1000, 1000000), chunks=(100, 1000000))

# Calculate and compute the mean
mean_dask = x_dask.mean().compute()
print(f"Dask Mean: {mean_dask}")
```

**Q5: Numba compilation timing**

    Numba uses Just-In-Time (JIT) compilation. The very first time you call a Numba-decorated function, the compiler has to translate the Python code into highly optimized machine code. This first call is very slow because of the compilation overhead.

    Therefore, when benchmarking, we exclude the first call so that we only measure the execution speed of the *compiled* code. In production, a server will compile the function once and then run it millions of times, so the compiled speed is the realistic speed to measure.
