### Question 1. Use gradient descent to minimize  𝑓(𝑥) = (𝑥 − 10)2.

```python
import numpy as np

def f(x):
    return (x - 10) ** 2

def grad_f(x):
    return 2 * (x - 10)

x = 0.0
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    x = x - learning_rate * grad_f(x)

print(f"Optimal x: {x:.4f}")
print(f"Minimum value: {f(x):.4f}")
```

### 2. Compare BFGS and CG on 𝑓(𝑥,𝑦) = (𝑥 − 4)2 +(𝑦 −6)2.

```python
from scipy.optimize import minimize

def f2(v):
    x, y = v
    return (x - 4)**2 + (y - 6)**2

x0 = [0, 0]

res_bfgs = minimize(f2, x0, method='BFGS')
print("--- BFGS ---")
print("Optimal x:", res_bfgs.x)
print("Iterations:", res_bfgs.nit)

res_cg = minimize(f2, x0, method='CG')
print("\n--- CG ---")
print("Optimal x:", res_cg.x)
print("Iterations:", res_cg.nit)
```

### 3. Explain the effect of changing the learning rate. 

**Q3: Effect of changing the learning rate**
* **Too small:** The algorithm takes tiny steps and convergence is very slow.
* **Too large:** The algorithm can overshoot the minimum and may diverge.
* **Just right:** The algorithm converges quickly and stably.

### 4. Use SciPy to estimate the mean of normally distributed observations using MLE.

```python
from scipy.optimize import minimize
import numpy as np

np.random.seed(42)
data = np.random.normal(loc=5, scale=2, size=100)

def neg_log_likelihood(mu):
    return -np.sum(-0.5 * np.log(2 * np.pi * 4) - 0.5 * ((data - mu) ** 2) / 4)

res = minimize(neg_log_likelihood, x0=0)
print(f"MLE estimate of mean: {res.x[0]:.4f}")
print(f"Sample mean: {np.mean(data):.4f}")
```

### 5. Implement a sigmoid function.

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("sigmoid(0):", sigmoid(0))
print("sigmoid(2):", sigmoid(2))
print("sigmoid(-2):", sigmoid(-2))
```

### Questions 6
A lecturer wants to predict the performance of a student (Y = 0 fail and Y = 1 pass) using 
the number of studying hours X. Data from 8 students is as shown below. 

X = [1, 2, 3, 4, 5, 6, 7, 8]
Y = [0, 0, 0, 0, 1, 1, 1, 1]

Fit a simple logistic regression model using MLE. 

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
Y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

model = LogisticRegression()
model.fit(X, Y)

print("Intercept (beta0):", model.intercept_[0])
print("Coefficient (beta1):", model.coef_[0][0])

for hours in [1, 4, 5, 8]:
    prob = model.predict_proba([[hours]])[0][1]
    print(f"Hours = {hours}: P(Pass) = {prob:.3f}")
```

**Q6: Logistic regression interpretation**
The model fits a sigmoid curve to the data. The intercept is approximately -5.29 and the coefficient is approximately 1.31. This means the log-odds of passing increase by 1.31 for each additional study hour. Students studying 5+ hours have a >60% chance of passing.
