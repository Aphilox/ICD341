### PART A: Basic Vectorization

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])

# Instead of a loop, we use vectorization:
y = 2 * x + 5

print("x:", x)
print("y = 2x + 5:", y)
```

### PART B: Vectorized Statistical Transformation

```python
marks = np.array([40, 55, 62, 70, 81, 90])

# Add 5 marks to every student
adjusted = marks + 5

# Convert marks to percentage
percentage = marks / 100

# Standardize the marks (z-scores)
z = (marks - np.mean(marks)) / np.std(marks)

print("Adjusted marks:", adjusted)
print("Percentage:", percentage)
print("Standardized z-scores:", z)
print("Mean of z:", z.mean())
```

### PART C: Broadcasting 

```python
data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])

# A 1x3 array (one value per column)
adjustment = np.array([5, 0, 2])

# Adding a 1x3 array to a 3x3 array
result = data + adjustment

print("Data:\n", data)
print("\nAdjustment:", adjustment)
print("\nResult:\n", result)
```

### PART D: Broadcasting With Statistics

```python
data = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

# Calculate the mean of each column
means = data.mean(axis=0)

# Subtract the column means from the data
centered = data - means

print("Original Data:\n", data)
print("\nColumn Means:", means)
print("\nCentered Data:\n", centered)

# Verify the mean of each column is now 0
print("\nMean of Centered Data (axis=0):", centered.mean(axis=0))
```

### PART E: Vectorization vs Loops

```python
# Create an array of 1,000,000 elements
x = np.arange(1_000_000)

# 1. Loop-based approach
result_loop = []
for value in x:
    result_loop.append(value**2 + 2*value + 1)

# 2. NumPy Vectorized approach
result_vectorized = x**2 + 2*x + 1

# Check if they agree
print("Are the two results equal?", np.allclose(result_loop, result_vectorized))
```

### PART F: Statistical Computing Application

```python
np.random.seed(123)
x = np.random.normal(100, 15, 100_000)

z = (x - x.mean()) / x.std()

print("Mean of z:", z.mean())
print("Std of z:", z.std())
```

### LAB 3 Challenge

```python
def standardize(x):
    return (x - np.mean(x)) / np.std(x)

a = np.array([1, 2, 3, 4, 5])
b = np.random.normal(10, 2, 100)
c = np.array([-5, 0, 5, 10])

print("Array 1:", standardize(a))
print("Array 2 (first 5):", standardize(b)[:5])
print("Array 3:", standardize(c))

# Modified for 2D datasets
def standardize_columns(x):
    return (x - np.mean(x, axis=0)) / np.std(x, axis=0)

data_2d = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

print("\nStandardized 2D Data:\n", standardize_columns(data_2d))
```

# Lab 3 Answers

**Q1: What is vectorization?**
Applying operations to whole arrays without explicit Python loops.

**Q2: Why is vectorization useful?**
Faster, shorter, less error-prone, and uses optimized C code.

**Q3: Difference between a Python loop and a NumPy vectorized operation?**
A Python loop runs element by element in Python. A NumPy vectorized operation runs compiled C loops over the whole array.

**Q4: Verify that the mean of z is approximately zero.**
The mean of z prints as a very small number (e.g., `1.85e-17`), which is essentially zero due to floating-point rounding.

**Q5: Why does broadcasting work for a 3x3 and 1x3 array?**
NumPy aligns dimensions from right to left. The 1x3 array is treated as (1, 3) and stretched across the 3 rows.

**Q6: Why is subtracting the vector means from the matrix possible?**
Broadcasting treats the means array (3,) as (1, 3) and subtracts column means from each column.

**Q7: Are the two results equal?** Yes.

**Q8: Which approach is easier to read?** Vectorized.

**Q9: Which is faster?** NumPy vectorized.

**Q10: Why?** Python loops have interpreter overhead; NumPy uses optimized C code.

**Q11: Why is vectorization useful for large statistical datasets?**
It avoids Python loop overhead and uses compiled C code, making it orders of magnitude faster for millions of data points.
