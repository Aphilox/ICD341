### Questions 1.
Solve      
	3𝑥+2𝑦 =12 
	𝑥 +4𝑦 = 10 
Using NumPy.

```python
import numpy as np

# 3x + 2y = 12
# x + 4y = 10
A = np.array([[3, 2], [1,4]])
b = np.array([12, 10])

solution = np.linalg.solve(A, b)
print("x =", solution[0])
print("y = ", solution[1])
```

### Questions 3.
 Explain why np.linalg.solve(A, b) is generally preferred to explicitly calculating 𝐴−1𝑏.

Using `np.linalg.solve()` is generally preferred because:

1. **Numerical stability:** Calculating the inverse directly can introduce large rounding errors, especially for ill-conditioned matrices. `solve()` uses more stable algorithms like LU decomposition with partial pivoting.
2. **Speed:** Inverting a matrix is roughly 2–3 times slower than solving the system directly.
3. **Memory:** Computing the full inverse requires storing an \(n \times n\) matrix, while `solve()` works directly on \(A\) and \(b\).

### Questions 4.
The data below relate to consumption expenditure and disposable income for a random 
sample of 10 households in Lusaka. 

Disposable Income X = [30, 45, 60, 100, 150, 200, 240, 250, 300, 320]
Consumption Y = [25, 20, 40, 65, 120, 190, 200, 200, 260, 220]

(i) Use a regression design matrix to estimate regression coefficients. 

```python
# Data
X = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])
Y = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# Build design matrix with intercept column
X_design = np.column_stack([np.ones(len(X)), X])

# Solve normal equations: beta = (X'X)^-1 X'Y  (solve is preferred)
beta = np.linalg.solve(X_design.T @ X_design, X_design.T @ Y)

print("Intercept (beta0):", beta[0])
print("Slope (beta1):", beta[1])
```

### Questions 4.
ii) Estimate the consumption for a household with disposable income of 181.

```python
income = 181
predicted = beta[0] + beta[1] * income
print(f"Predicted consumption for income {income}: {predicted:.2f}")
```
