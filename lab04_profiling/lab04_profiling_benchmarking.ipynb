```python
import numpy as np

x = np.arange(1_000_000)

# Benchmark the vectorized operation
%timeit x ** 2
```

```python
# 1. Define the loop-based function
def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result

# 2. Define the NumPy vectorized function
def square_numpy(x):
    return x ** 2

# 3. Benchmark both
%timeit square_loop(x)
%timeit square_numpy(x)
```

```python
# Set the seed for reproducibility
np.random.seed(42)

# Generate 5,000,000 observations from a normal distribution
data = np.random.normal(
    loc=100,
    scale=20,
    size=5_000_000
)

# Approach 1: Loop-based
def squared_deviations_loop(x):
    mean = np.mean(x)
    total = 0
    for value in x:
        total += (value - mean) ** 2
    return total

# Approach 2: NumPy Vectorized
def squared_deviations_numpy(x):
    mean = np.mean(x)
    return np.sum((x - mean) ** 2)

# Benchmark both approaches
print("Benchmarking Loop Approach...")
%timeit squared_deviations_loop(data)

print("\nBenchmarking NumPy Vectorized Approach...")
%timeit squared_deviations_numpy(data)
```
