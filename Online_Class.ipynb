```python
import numpy as np 
A = np.array([ 
[1, 2], 
[3, 4] 
]) 
Q, R = np.linalg.qr(A) 
print("Q:") 
print(Q) 
print("R:") 
print(R) 
#Verify: 
print(np.allclose(Q @ R, A))
```

```python
X = np.array([ 
[1, 1], 
[1, 2], 
[1, 3], 
[1, 4] 
]) 
y = np.array([3, 5, 7, 9]) 
Q, R = np.linalg.qr(X) 
beta = np.linalg.solve(R, Q.T @ y) 
print(beta)
```

```python
beta_normal = np.linalg.solve(X.T @ X, X.T @ y) 
Q, R = np.linalg.qr(X) 
beta_qr = np.linalg.solve(R, Q.T @ y) 
print(beta_normal) 
print(beta_qr)
```

```python
A = np.array([ 
[1, 2], 
[3, 4] 
])  
U, S, Vt = np.linalg.svd(A) 
print("U:") 
print(U) 
print("Singular values:") 
print(S) 
print("V transpose:") 
print(Vt)
```

```python
Sigma = np.diag(S) 
A_reconstructed = U @ Sigma @ Vt 
print(A_reconstructed) 
```

```python
A = np.array([ 
[1, 2], 
[2, 4] 
]) 
rank = np.linalg.matrix_rank(A) 
print(rank) 
```
